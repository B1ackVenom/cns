# Practical 1B: Vigenère Cipher

**Aim:** To implement polyalphabetic substitution encryption and decryption using the Vigenère Cipher.

### Implementation

```python
def encrypt(text, key):
    key = key.upper()
    result = ""
    i = 0

    for ch in text:
        if ch.isalpha():
            shift = ord(key[i % len(key)]) - ord('A')
            base = ord('A') if ch.isupper() else ord('a')
            result += chr((ord(ch) - base + shift) % 26 + base)
            i += 1
        else:
            result += ch

    return result


def decrypt(text, key):
    key = key.upper()
    result = ""
    i = 0

    for ch in text:
        if ch.isalpha():
            shift = ord(key[i % len(key)]) - ord('A')
            base = ord('A') if ch.isupper() else ord('a')
            result += chr((ord(ch) - base - shift) % 26 + base)
            i += 1
        else:
            result += ch

    return result


plaintext = input("Enter plaintext: ")
key = input("Enter key: ")

encrypted = encrypt(plaintext, key)
decrypted = decrypt(encrypted, key)

print("Encrypted:", encrypted)
print("Decrypted:", decrypted)
```
