# Practical 5: Cryptographic Hashing (MD5 & SHA-256)

**Aim:** To compute and compare MD5 and SHA-256 cryptographic message digests.

### Implementation

```python
import hashlib

msg = input("Message: ").encode()
print("MD5    :", hashlib.md5(msg).hexdigest())
print("SHA-256:", hashlib.sha256(msg).hexdigest())
```
